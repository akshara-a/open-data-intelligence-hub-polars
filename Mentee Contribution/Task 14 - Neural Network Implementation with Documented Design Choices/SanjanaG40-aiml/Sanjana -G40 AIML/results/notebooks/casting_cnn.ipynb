```python
import os
import tensorflow as tf

# ============================================================
# 1. Dataset paths
# ============================================================

PROJECT_DIR = r"C:\Users\22053\Programming-Language-Popularity-Tracker\open-data-intelligence-hub\Mentee Contribution\Task 14 - Neural Network Implementation with Documented Design Choices\SanjanaG40-aiml"

DATASET_DIR = os.path.join(
    PROJECT_DIR,
    "data",
    "archive (1)",
    "casting_data",
    "casting_data"
)

TRAIN_DIR = os.path.join(DATASET_DIR, "train")
TEST_DIR = os.path.join(DATASET_DIR, "test")

# ============================================================
# 2. Project settings
# ============================================================

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

print("Dataset directory:")
print(DATASET_DIR)

print("\nTrain directory exists:", os.path.exists(TRAIN_DIR))
print("Test directory exists:", os.path.exists(TEST_DIR))

# ============================================================
# 3. Load training dataset
#    80% of train data -> training
#    20% of train data -> validation
# ============================================================

train_dataset = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    validation_split=0.20,
    subset="training",
    seed=SEED,
    shuffle=True
)

# ============================================================
# 4. Load validation dataset
# ============================================================

validation_dataset = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    validation_split=0.20,
    subset="validation",
    seed=SEED,
    shuffle=True
)

# ============================================================
# 5. Load test dataset
#    Test data remains completely untouched
# ============================================================

test_dataset = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("\nClass names:")
print(train_dataset.class_names)

print("\nDatasets loaded successfully!")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
```

```python
model = keras.Sequential([
    layers.Rescaling(1./255, input_shape=(224, 224, 3)),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D(),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(1, activation='sigmoid')
])

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=10
)
```

```python
test_loss, test_accuracy = model.evaluate(test_dataset)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Accuracy (%):", test_accuracy * 100)
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array

# --------------------------------------------------
# 1. Set your test dataset path
# --------------------------------------------------
test_dir = r"C:\Users\22053\Programming-Language-Popularity-Tracker\open-data-intelligence-hub\Mentee Contribution\Task 14 - Neural Network Implementation with Documented Design Choices\SanjanaG40-aiml\data\archive (1)\casting_data\casting_data\test"

class_names = ['def_front', 'ok_front']

# --------------------------------------------------
# 2. Function to find images from each class
# --------------------------------------------------
def get_images(class_name):
    class_dir = os.path.join(test_dir, class_name)

    extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.webp')

    images = [
        os.path.join(class_dir, f)
        for f in os.listdir(class_dir)
        if f.lower().endswith(extensions)
    ]

    return images


# --------------------------------------------------
# 3. Pick one random image from each class
# --------------------------------------------------
selected_images = []

for class_name in class_names:
    images = get_images(class_name)

    if len(images) == 0:
        print(f"No images found for {class_name}")
    else:
        selected_images.append(
            (random.choice(images), class_name)
        )


# --------------------------------------------------
# 4. Predict and display the images
# --------------------------------------------------
plt.figure(figsize=(10, 5))

for i, (image_path, actual_class) in enumerate(selected_images):

    # Load image
    img = load_img(image_path, target_size=(224, 224))
    img_array = img_to_array(img)

    # Normalize exactly like the model
    img_input = np.expand_dims(img_array, axis=0)

    # Model prediction
    probability = model.predict(img_input, verbose=0)[0][0]

    # Sigmoid output:
    # < 0.5 = def_front
    # >= 0.5 = ok_front
    if probability >= 0.5:
        predicted_class = 'ok_front'
        confidence = probability
    else:
        predicted_class = 'def_front'
        confidence = 1 - probability

    # Display
    plt.subplot(1, 2, i + 1)
    plt.imshow(img)
    plt.axis('off')

    plt.title(
        f"Actual: {actual_class}\n"
        f"Predicted: {predicted_class}\n"
        f"Confidence: {confidence * 100:.2f}%"
    )

plt.tight_layout()
plt.show()
```

# Neural Network Design Choices and Results

## 1. Problem Definition

This project performs binary image classification of casting images into two classes:

- `def_front` – defective casting
- `ok_front` – acceptable casting

A Convolutional Neural Network (CNN) was selected because CNNs are well suited for image classification and can automatically learn visual features such as edges, shapes, textures, and defect patterns.

## 2. Dataset

The dataset contains two classes:

- `def_front`
- `ok_front`

The dataset was divided into training, validation, and testing sets.

- Training images: 5,307
- Validation images: 1,326
- Test images: 715

The test set was kept separate from training so that the final model performance could be evaluated on unseen images.

## 3. Image Preprocessing

Images were resized to 224 × 224 pixels with 3 color channels.

Pixel values were normalized to the range 0–1 using a Rescaling layer with a factor of 1/255.

This normalization helps the neural network train more efficiently.

## 4. CNN Architecture

The implemented CNN contains three convolutional blocks.

Each convolutional block consists of:

- Convolutional layer
- ReLU activation
- MaxPooling layer

The convolutional layers use 32, 64, and 128 filters respectively. This allows the network to learn increasingly complex image features.

After convolution and pooling, the feature maps are flattened and passed through:

- Dense layer with 128 neurons
- ReLU activation
- Dropout with a rate of 0.5
- Output layer with one neuron
- Sigmoid activation

The sigmoid output is appropriate because this is a binary classification problem.

## 5. Training Configuration

The model was compiled using:

- Optimizer: Adam
- Loss function: Binary Cross-Entropy
- Evaluation metric: Accuracy
- Number of epochs: 10

Adam was selected because it provides adaptive learning rates and generally performs well for neural-network training.

Binary cross-entropy was selected because the task contains two classes.

## 6. Model Evaluation

The trained model achieved:

- Test Loss: 0.0440
- Test Accuracy: 98.32%

The model correctly classified 703 out of 715 test images.

Only 12 images were incorrectly classified.

## 7. Confusion Matrix

The confusion matrix showed:

- 442 `def_front` images correctly classified
- 261 `ok_front` images correctly classified
- 11 `def_front` images classified as `ok_front`
- 1 `ok_front` image classified as `def_front`

This demonstrates that the model performs well on both classes.

## 8. Classification Report

The model achieved the following results:

| Class | Precision | Recall | F1-score |
|-------|-----------|--------|----------|
| def_front | 1.00 | 0.98 | 0.99 |
| ok_front | 0.96 | 1.00 | 0.98 |

The high precision, recall, and F1-scores indicate strong classification performance.

## 9. Individual Image Testing

Individual unseen test images were also passed through the trained model.

Example results:

- `def_front` → `def_front`, confidence 92.30%
- `ok_front` → `ok_front`, confidence 100.00%

Both sample images were correctly classified.

## 10. Model Saving

The trained model was saved as:

`casting_cnn_model.keras`

This allows the trained neural network to be loaded and reused without retraining.

## 11. Conclusion

The implemented Convolutional Neural Network successfully classified defective and acceptable casting images. The model achieved a test accuracy of 98.32%, demonstrating strong performance on unseen test data.

The confusion matrix and classification report further confirmed that the model performs consistently across both classes. Individual image testing also produced correct predictions with high confidence.

Therefore, the implemented CNN provides an effective solution for automated casting image classification.
