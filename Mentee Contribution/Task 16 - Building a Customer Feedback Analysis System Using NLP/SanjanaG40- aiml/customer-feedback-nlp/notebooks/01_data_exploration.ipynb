# 01 - Data Exploration

In this notebook we load and explore the dataset before doing any modeling. Understanding the data is the first and most important step in any NLP project.

## Dataset: Twitter US Airline Sentiment

The dataset contains ~14,640 tweets about six US airlines, crowd-labeled with sentiment (`airline_sentiment`) and, for negative tweets, a complaint reason (`negativereason`).

The CSV is created by running `python make_dataset.py` (or downloaded into `data/raw/Tweets.csv`).

```python
import sys
sys.path.insert(0, "..")  # allow importing the src package

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_colwidth", 120)
```

```python
from src.data_loader import load_raw_tweets

df = load_raw_tweets()
print(f"Shape: {df.shape}")
df.head()
```

## Columns

- `text` - the tweet text (our feedback)
- `airline_sentiment` - positive / neutral / negative (crowd label)
- `negativereason` - complaint reason (only for negative tweets)
- `airline` - which airline the tweet is about

```python
print("Column data types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isna().sum())
```

## Sentiment distribution

```python
sns.countplot(data=df, x="airline_sentiment", order=["negative", "neutral", "positive"])
plt.title("Sentiment distribution")
plt.show()

print(df["airline_sentiment"].value_counts(normalize=True).round(3))
```

Notice the class imbalance: **negative tweets dominate** (about 63%). This matters for evaluation.

## Complaint reasons (negativereason)

Only negative tweets have a reason. This is our **category** signal, but it is airline-specific and does not match the project's app-feedback categories (payment, login, support, ...). See the README/labeling strategy for how we handle this.

```python
sns.countplot(y=df["negativereason"].dropna(), order=df["negativereason"].dropna().value_counts().index)
plt.title("Complaint reasons")
plt.tight_layout()
plt.show()
```

## Tweet text length analysis

```python
df["text_length"] = df["text"].str.len()
print(df["text_length"].describe())

sns.histplot(df["text_length"], bins=50)
plt.title("Distribution of tweet text length")
plt.show()
```

## Example tweets per sentiment

```python
for sent in ["negative", "neutral", "positive"]:
    sample = df[df["airline_sentiment"] == sent]["text"].sample(2, random_state=42)
    print(f"--- {sent} ---")
    for t in sample:
        print("  ", t[:140])
    print()
```
