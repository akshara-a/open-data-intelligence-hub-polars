```python
import pandas as pd
import numpy as np
import os

# -----------------------------
# Set random seed
# -----------------------------
np.random.seed(42)

# Number of customers
n = 5000

# -----------------------------
# Generate customer data
# -----------------------------
customer_id = np.arange(100001, 100001 + n)

age = np.random.randint(18, 70, n)

gender = np.random.choice(
    ["Male", "Female"],
    size=n,
    p=[0.5, 0.5]
)

location = np.random.choice(
    ["North", "South", "East", "West"],
    size=n
)

device_type = np.random.choice(
    ["Mobile", "Desktop", "Tablet"],
    size=n,
    p=[0.6, 0.3, 0.1]
)

traffic_source = np.random.choice(
    ["Search", "Social Media", "Email", "Direct", "Advertisement"],
    size=n
)

pages_viewed = np.random.randint(1, 30, n)

time_on_site = np.random.randint(30, 1200, n)

products_viewed = np.random.randint(1, 20, n)

cart_items = np.random.randint(0, 10, n)

previous_purchases = np.random.randint(0, 25, n)

average_order_value = np.round(
    np.random.uniform(10, 500, n),
    2
)

discount_used = np.random.choice(
    [0, 1],
    size=n,
    p=[0.6, 0.4]
)

email_clicked = np.random.choice(
    [0, 1],
    size=n,
    p=[0.5, 0.5]
)

ad_clicked = np.random.choice(
    [0, 1],
    size=n,
    p=[0.7, 0.3]
)

review_score_viewed = np.round(
    np.random.uniform(1, 5, n),
    1
)

days_since_last_visit = np.random.randint(0, 60, n)

session_count = np.random.randint(1, 20, n)

# -----------------------------
# Purchase Score
# -----------------------------
purchase_score = (
    0.05 * pages_viewed +
    0.003 * time_on_site +
    0.25 * cart_items +
    0.08 * previous_purchases +
    0.15 * discount_used +
    0.10 * email_clicked +
    0.05 * ad_clicked +
    0.04 * review_score_viewed -
    0.03 * days_since_last_visit +
    np.random.normal(0, 1, n)
)

purchase = (
    purchase_score > np.percentile(purchase_score, 55)
).astype(int)

# -----------------------------
# Create DataFrame
# -----------------------------
df = pd.DataFrame({
    "CustomerID": customer_id,
    "Age": age,
    "Gender": gender,
    "Location": location,
    "DeviceType": device_type,
    "TrafficSource": traffic_source,
    "PagesViewed": pages_viewed,
    "TimeOnSite": time_on_site,
    "ProductsViewed": products_viewed,
    "CartItems": cart_items,
    "PreviousPurchases": previous_purchases,
    "AverageOrderValue": average_order_value,
    "DiscountUsed": discount_used,
    "EmailClicked": email_clicked,
    "AdClicked": ad_clicked,
    "ReviewScoreViewed": review_score_viewed,
    "DaysSinceLastVisit": days_since_last_visit,
    "SessionCount": session_count,
    "Purchase": purchase
})

# -----------------------------
# Save Dataset
# -----------------------------
output_path = os.path.join("..", "data", "ecommerce_customer_data.csv")

df.to_csv(output_path, index=False)

print("=" * 50)
print("Dataset Generated Successfully!")
print("=" * 50)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nTarget Distribution:")
print(df["Purchase"].value_counts())

print(f"\nDataset saved to:\n{output_path}")
```
