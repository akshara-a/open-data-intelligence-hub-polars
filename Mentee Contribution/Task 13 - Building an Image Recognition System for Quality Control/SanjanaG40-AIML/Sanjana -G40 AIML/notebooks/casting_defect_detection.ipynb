```python
import os
import shutil
import random

# --------------------------------------------------
# 1. Source dataset
# --------------------------------------------------

source_directory = r"C:\Users\22053\Programming-Language-Popularity-Tracker\open-data-intelligence-hub\Mentee Contribution\Task 13 - Building an Image Recognition System for Quality Control\SanjanaG40-AIML\data\archive (1)\casting_512x512\casting_512x512"

# --------------------------------------------------
# 2. Destination directory
# --------------------------------------------------

output_directory = r"C:\Users\22053\Programming-Language-Popularity-Tracker\open-data-intelligence-hub\Mentee Contribution\Task 13 - Building an Image Recognition System for Quality Control\SanjanaG40-AIML\data\processed"

# --------------------------------------------------
# 3. Classes
# --------------------------------------------------

classes = ["ok_front", "def_front"]

# --------------------------------------------------
# 4. Split ratios
# --------------------------------------------------

train_ratio = 0.70
validation_ratio = 0.15
test_ratio = 0.15

# Make sure ratios add up to 1
assert train_ratio + validation_ratio + test_ratio == 1.0

# --------------------------------------------------
# 5. Reproducibility
# --------------------------------------------------

random.seed(42)

# --------------------------------------------------
# 6. Create destination folders
# --------------------------------------------------

for split in ["train", "validation", "test"]:
    for class_name in classes:
        folder = os.path.join(output_directory, split, class_name)
        os.makedirs(folder, exist_ok=True)

# --------------------------------------------------
# 7. Split and copy images
# --------------------------------------------------

for class_name in classes:

    source_class_directory = os.path.join(
        source_directory,
        class_name
    )

    images = [
        file for file in os.listdir(source_class_directory)
        if os.path.isfile(
            os.path.join(source_class_directory, file)
        )
    ]

    # Shuffle images
    random.shuffle(images)

    total_images = len(images)

    train_end = int(total_images * train_ratio)
    validation_end = train_end + int(
        total_images * validation_ratio
    )

    train_images = images[:train_end]
    validation_images = images[train_end:validation_end]
    test_images = images[validation_end:]

    # Copy training images
    for image in train_images:
        source = os.path.join(
            source_class_directory,
            image
        )

        destination = os.path.join(
            output_directory,
            "train",
            class_name,
            image
        )

        shutil.copy2(source, destination)

    # Copy validation images
    for image in validation_images:
        source = os.path.join(
            source_class_directory,
            image
        )

        destination = os.path.join(
            output_directory,
            "validation",
            class_name,
            image
        )

        shutil.copy2(source, destination)

    # Copy test images
    for image in test_images:
        source = os.path.join(
            source_class_directory,
            image
        )

        destination = os.path.join(
            output_directory,
            "test",
            class_name,
            image
        )

        shutil.copy2(source, destination)

    print(f"\n{class_name}")
    print(f"Total:      {total_images}")
    print(f"Training:   {len(train_images)}")
    print(f"Validation: {len(validation_images)}")
    print(f"Testing:    {len(test_images)}")
```

```python
import os

processed_directory = r"C:\Users\22053\Programming-Language-Popularity-Tracker\open-data-intelligence-hub\Mentee Contribution\Task 13 - Building an Image Recognition System for Quality Control\SanjanaG40-AIML\data\processed"

train_directory = os.path.join(processed_directory, "train")
validation_directory = os.path.join(processed_directory, "validation")
test_directory = os.path.join(processed_directory, "test")

print("Train:", train_directory)
print("Validation:", validation_directory)
print("Test:", test_directory)
```

```python
import tensorflow as tf

image_size = (224, 224)
batch_size = 32

class_names = ["ok_front", "def_front"]

train_dataset = tf.keras.utils.image_dataset_from_directory(
    train_directory,
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="binary",
    shuffle=True,
    seed=42
)

validation_dataset = tf.keras.utils.image_dataset_from_directory(
    validation_directory,
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="binary",
    shuffle=False
)

test_dataset = tf.keras.utils.image_dataset_from_directory(
    test_directory,
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="binary",
    shuffle=False
)
```

```python
print("Class names:", class_names)

for images, labels in train_dataset.take(1):
    print("Image batch shape:", images.shape)
    print("Label batch shape:", labels.shape)
    print("First 10 labels:", labels.numpy().flatten()[:10])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

for images, labels in train_dataset.take(1):

    for i in range(9):

        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i].numpy().astype("uint8"))

        label = int(labels[i].numpy()[0])

        if label == 0:
            title = "Non-defective"
        else:
            title = "Defective"

        plt.title(title)
        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_dataset = train_dataset.prefetch(AUTOTUNE)
validation_dataset = validation_dataset.prefetch(AUTOTUNE)
test_dataset = test_dataset.prefetch(AUTOTUNE)
```

```python
from tensorflow.keras import layers

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10),
    layers.RandomTranslation(0.05, 0.05),
    layers.RandomContrast(0.10)
], name="data_augmentation")

print(data_augmentation)
```

```python
plt.figure(figsize=(10, 8))

for images, labels in train_dataset.take(1):

    original_image = images[0]

    for i in range(9):
        augmented_image = data_augmentation(
            tf.expand_dims(original_image, 0),
            training=True
        )[0]

        plt.subplot(3, 3, i + 1)
        plt.imshow(
            tf.cast(
                tf.clip_by_value(augmented_image, 0, 255),
                tf.uint8
            )
        )

        label = int(labels[0].numpy()[0])

        plt.title(
            "Non-defective" if label == 0 else "Defective"
        )
        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    
    # Input
    layers.Input(shape=(224, 224, 3)),

    # Data augmentation
    data_augmentation,

    # Normalize pixel values from 0-255 to 0-1
    layers.Rescaling(1.0 / 255),

    # Convolution Block 1
    layers.Conv2D(32, kernel_size=3, activation="relu"),
    layers.MaxPooling2D(),

    # Convolution Block 2
    layers.Conv2D(64, kernel_size=3, activation="relu"),
    layers.MaxPooling2D(),

    # Convolution Block 3
    layers.Conv2D(128, kernel_size=3, activation="relu"),
    layers.MaxPooling2D(),

    # Feature reduction
    layers.GlobalAveragePooling2D(),

    # Regularization
    layers.Dropout(0.40),

    # Fully connected layer
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.30),

    # Binary output
    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)
```
